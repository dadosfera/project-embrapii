```python
from src.utilitis import *
import pandas as pd
```

```python
models = ["Qwen2.5-Coder-32B-Instruct",  "Qwen2.5-Coder-14B-Instruct", "Qwen2.5-Coder-7B-Instruct", "Llama-3.1-8B-Instruct", "llama-3-sqlcoder-8b"]
bibliotecas = ["PremSQLAgente", "vannaAi", "vannaAi_contexto"]
```

```python
db_name = "sih_database"
print(f"EX - {db_name}")
data_path = f"resources/out/{db_name}/"
print("Model", end='\t')
for biblioteca in bibliotecas:
    print(f"{biblioteca}", end='\t')
print()

for model_name in models:
    print(f"{model_name}", end='\t')
    for biblioteca in bibliotecas:
        model_id = get_model_id(model_name)
        data_file = f"{data_path}/{model_id}/queries_geradas_{biblioteca}_executado.parquet"
        try:
            df = pd.read_parquet(data_file)
            print(f"{df['execucoes_iguais'].value_counts(normalize=True)[True]:.2f}", end='\t')
        except:
            print(f"-", end='\t')
    print()
```
